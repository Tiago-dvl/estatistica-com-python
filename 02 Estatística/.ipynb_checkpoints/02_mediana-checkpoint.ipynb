```python
dados = [12, 45, 8, 23, 15, 31, 19]

tempos = sorted(dados)
formula_mediana = (int((len(tempos) + 1) / 2)) -1
mediana_tempos = tempos[formula_mediana]

print(f"O conjunto de dados é: ", tempos)
print(f"A mediana é: ",mediana_tempos)
```

## Kata: Mediana com statistics.median()
Descrição do objetivo:
Você tem uma lista com o valor (em R$) de 8 pedidos feitos em uma loja online:

python
pedidos = [120, 85, 310, 45, 200, 95, 150, 60]

Escreva um programa que encontre a mediana desses valores usando statistics.median() e mostre o resultado na tela com o texto explicativo. Repare que a lista tem uma quantidade par de valores.

Resultado final esperado:

A mediana dos pedidos é: 110.0

```python
# metodo sem statistics

df = sorted([120, 85, 310, 45, 200, 95, 150, 60])

x_01 = int((len(df) / 2) -1)
x = df[x_01]

y_01 = int(((len(df) / 2) + 1) -1)
y = df[y_01]
z = (x + y) / 2

print(f"A mediana dos pedidos é: ",z)



```

```python
# método com statistics

import statistics

python_pedidos = [120, 85, 310, 45, 200, 95, 150, 60]

df = statistics.median(python_pedidos)

print(f"A mediana dos pedidos é: ",df)
```
